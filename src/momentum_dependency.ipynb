# Momentum dependence of loss both non-Hermitian and Lindbladian

- Lindbladian
$$\frac{d\rho}{dt}=-\frac{i}{\hbar}[H_\text{free}, \rho] + \sum_p \left(L_p\rho L_p^\dagger - \frac{1}{2}\{L_p^\dagger L_p, \rho\}\right)$$
$$L_p=\sqrt{\frac{\gamma}{V}}\sum_q c_\uparrow(q-k)c_\downarrow(k)$$
multiplying $n_\sigma(q)$ to the left
$$
\frac{d}{dt}\rho n_\sigma(q_x)=-\frac{i}{\hbar}[H_\text{free}\rho n_\sigma(q_x)-\rho H_\text{free}n_\sigma(q_x)]+\sum_p\left(L_p\rho L_p^\dagger n_\sigma(q_x)-\frac{1}{2}L_p^\dagger L_p\rho n_\sigma(q_x)-\frac{1}{2}\rho L_p^\dagger L_p n_\sigma(q_x)\right)
$$
apply trace on both side and rearrange:
$$
\frac{d}{dt}\braket{n_\sigma(q_x)}=-\frac{i}{\hbar}\underbrace{\braket{[n_\sigma(q_x), H_\text{free}]}}_{0}+\left\langle\sum_p\left(\frac{1}{2}L_p^\dagger n_\sigma(q_x)L_p-\frac{1}{2} L_p^\dagger L_pn_\sigma(q_x)\right)+\underbrace{\sum_p \left(\frac{1}{2}L_p^\dagger n_\sigma(q_x)L_p-\frac{1}{2} n_\sigma(q_x)L_p^\dagger L_p\right)}_{\text{h.c.}}\right\rangle
$$
$$
\begin{align}
	\frac{d}{dt}\braket{n_\sigma(q_x)}&=\frac{1}{2}\sum_p\left\langle L_p^\dagger[n_\sigma(q_x), L_p]\right\rangle+\text{c.c.} \\
	&=\sqrt{\frac{\gamma}{V}}\sum_p\operatorname{Re}\left\langle L_p^\dagger\left[c_\sigma(q_x)^\dagger c_\sigma(q_x), \sum_{p'}c_\uparrow(p-p')c_\downarrow(p')\right]\right\rangle
\end{align}
$$
for $\sigma=\uparrow$:
$$
\begin{align}
	\frac{d}{dt}\braket{n_\uparrow(q_x)}&=\sqrt{\frac{\gamma}{V}}\sum_p\operatorname{Re}\left\langle L_p^\dagger\sum_{p'}\left[c_\uparrow(q_x)^\dagger c_\uparrow(q_x), c_\uparrow(p-p')\right]c_\downarrow(p')\right\rangle\\
	&=-\sqrt{\frac{\gamma}{V}}\sum_p\operatorname{Re}\left\langle L_p^\dagger c_\uparrow(q_x)c_\downarrow(p-q_x)\right\rangle\\
	&=-\frac{\gamma}{V}\sum_{p,p'}\operatorname{Re}\left(c_\downarrow^\dagger(p-p')c_\uparrow^\dagger(p') c_\uparrow(q_x)c_\downarrow(p-q_x)\right\rangle
\end{align}
$$
similarly, for $\sigma=\downarrow$,
$$
\begin{align}
	\frac{d}{dt}\braket{n_\downarrow(q_x)}&=\sqrt{\frac{\gamma}{V}}\sum_p\operatorname{Re}\left\langle L_p^\dagger\sum_{p'}c_\uparrow(p-p')\left[c_\downarrow(q_x)^\dagger c_\downarrow(q_x), c_\downarrow(p')\right]\right\rangle\\
	&=-\sqrt{\frac{\gamma}{V}}\sum_p\operatorname{Re}\left\langle L_p^\dagger c_\uparrow(p-q_x) c_\downarrow(q_x)\right\rangle\\
	&=-\frac{\gamma}{V}\sum_{p,p'}\operatorname{Re}\left(c_\downarrow^\dagger(p-p')c_\uparrow^\dagger(p') c_\uparrow(p-q_x) c_\downarrow(q_x)\right\rangle
\end{align}
$$
- Non-Hermitian
ignore the jump term:
$$
\frac{d}{dt}\rho n_\sigma(q_x)=-\frac{i}{\hbar}[H_\text{free}\rho n_\sigma(q_x)-\rho H_\text{free}n_\sigma(q_x)]-\frac{1}{2}\sum_p\left(L_p^\dagger L_p\rho n_\sigma(q_x)+\rho L_p^\dagger L_p n_\sigma(q_x)\right)$$
take trace, cyclic rearrange
$$
\begin{align}
\frac{d}{dt}\braket{n_\sigma(q_x)}&=-\frac{i}{\hbar}\underbrace{\braket{[n_\sigma(q_x), H_\text{free}]}}_{0}-\frac{1}{2}\sum_p\left(\braket{n_\sigma(q_x)L_p^\dagger L_p}+\braket{L_p^\dagger L_p n_\sigma(q_x)}\right)\\
&=-\frac{i}{\hbar}\underbrace{\braket{[n_\sigma(q_x), H_\text{free}]}}_{0}-\sum_p\braket{n_\sigma(q_x)L_p^\dagger L_p}\\
&=-\frac{\gamma}{V}\sum_{p,p_1,p_2}\left\langle{c_\sigma^\dagger(q_x)c_\sigma(q_x)c_\downarrow^\dagger(p_2-p)c_\uparrow^\dagger(p_1+p)}c_\uparrow(p_1)c_\downarrow(p_2)\right\rangle
\end{align}
$$

Hence, the short time evolution is governed by:
- Lindbladian
$$
\frac{d\braket{n_\uparrow(q_x)}}{dt} = -\frac{\gamma}{V}\sum_{p, p'} \operatorname{Re}\left\langle c_\downarrow^\dagger(p-p')c_\uparrow^\dagger(p')c_\uparrow(q_x)c_\downarrow(p-q_x)\right\rangle
$$
$$
\frac{d\braket{n_\downarrow(q_x)}}{dt} = -\frac{\gamma}{V}\sum_{p, p'} \operatorname{Re}\left\langle c_\downarrow^\dagger(p-p')c_\uparrow^\dagger(p')c_\uparrow(p-q_x)c_\downarrow(q_x)\right\rangle
$$
$$
\left.\frac{d}{dt}\braket{n_\uparrow(q_x)}\right|_{t=0} = -\sum_p \left[\operatorname{Re}\left\langle c_\downarrow^\dagger(p)c_\uparrow^\dagger(q_x)c_\uparrow(q_x)c_\downarrow(p) \right\rangle _F + \operatorname{Re}\left\langle c_\downarrow^\dagger(q_x)c_\uparrow^\dagger(p)c_\uparrow(q_x)c_\downarrow(p) \right\rangle _F\right]
$$


- non-Hermitian
$$
\frac{d\braket{n_\sigma(q_x)}}{dt} = -\frac{\gamma}{V}\sum_{p,p_1,p_2}\left\langle{c_\sigma^\dagger(q_x)c_\sigma(q_x)c_\downarrow^\dagger(p_2-p)c_\uparrow^\dagger(p_1+p)}c_\uparrow(p_1)c_\downarrow(p_2)\right\rangle
$$






```python
import os
os.environ["XLA_PYTHON_CLIENT_PREALLOCATE"] = "false"

import numpy as np
import matplotlib.pyplot as plt

from quant_mech import ybsoc

from matplotlib.collections import LineCollection
from matplotlib.colors import LinearSegmentedColormap, Normalize
from matplotlib.cm import ScalarMappable
```

```python
t_unit = 0.1129 # ms
E_unit = 9.3428e-31 # J  recoil energy
T_unit = 6.7669e-2 # μK

hbar = 1.0
k_r = 1.0
m_Yb = 1.0
delta = 4.0
omega_R = 3.5
gamma = 0.5
temperature = 0.02 / T_unit
mu = 2.0

fermi_level = mu

m = 100
q_left = -8
q_right = 6

zerotemp = temperature < 1e-12
```

```python
points_to_calc = np.linspace(q_left, q_right, (q_right - q_left) * m + 1)
int_momentums = np.linspace(q_left * m, q_right * m, (q_right - q_left) * m + 1, dtype=np.int64)
zero_idx = np.argmin(np.abs(points_to_calc))

k0 = points_to_calc[1] - points_to_calc[0]
L = 2 * np.pi / k0

energies = np.stack([
    ybsoc.e2(points_to_calc, m_Yb, k_r, delta, omega_R, hbar),
    ybsoc.e1(points_to_calc, m_Yb, k_r, delta, omega_R, hbar)
])
above_fermi = energies > fermi_level
below_fermi = energies <= fermi_level
if zerotemp:
    fermi_dirac = below_fermi.astype(float)
else:
    beta = 1 / temperature
    fermi_dirac = 1 / (np.exp(beta * (energies - mu)) + 1)


thetas = ybsoc.theta(points_to_calc, m_Yb, k_r, delta, omega_R, hbar)
alphas = np.cos(thetas)
betas = np.sin(thetas)
```

```python
qs_dense = np.linspace(-8, 6, 1401)

E_upper_dense = ybsoc.e1(qs_dense, m_Yb, k_r, delta, omega_R, hbar)
E_lower_dense = ybsoc.e2(qs_dense, m_Yb, k_r, delta, omega_R, hbar)

thetas_dense = ybsoc.theta(
    qs_dense,
    delta=delta,
    omega_R=omega_R,
    k_r=k_r,
    m_Yb=m_Yb,
)
alphas_dense = np.cos(thetas_dense)
betas_dense = np.sin(thetas_dense)

custom_cmap = LinearSegmentedColormap.from_list("custom_cmap", ['tomato', 'cornflowerblue'])

color_determining_array_up = np.abs(betas_dense)**2
color_determining_array_down = np.abs(alphas_dense)**2

points_up = np.array([qs_dense, E_upper_dense]).T.reshape(-1, 1, 2)
segments_up = np.concatenate([points_up[:-1], points_up[1:]], axis=1)

points_down = np.array([qs_dense, E_lower_dense]).T.reshape(-1, 1, 2)
segments_down = np.concatenate([points_down[:-1], points_down[1:]], axis=1)

norm = Normalize(vmin=0.0, vmax=1.0)
lc_up = LineCollection(segments_up, cmap=custom_cmap, norm=norm)
lc_down = LineCollection(segments_down, cmap=custom_cmap, norm=norm)

lc_up.set_array(color_determining_array_up[:-1])
lc_down.set_array(color_determining_array_down[:-1])
lc_up.set_linewidth(2)
lc_down.set_linewidth(2)

fig, ax = plt.subplots(figsize=(6, 4))

ax.add_collection(lc_up)
ax.add_collection(lc_down)

ax.set_xlim(qs_dense.min(), qs_dense.max())
ax.set_ylim(E_lower_dense.min() - 1, E_upper_dense.max() + 1)

sm = ScalarMappable(cmap=custom_cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ticks=[0, 1], ax=ax)
cbar.set_ticklabels([r'$m_F=5/2$', r'$m_F=3/2$'])
cbar.ax.invert_yaxis()
cbar.set_label(r'Mixing Ratio')
ax.set_xlabel("$q_x/k_r$")
ax.set_ylabel("$E/E_r$")
ax.grid(True)


if zerotemp:
    q_fermi_sea = np.concatenate([
        points_to_calc[below_fermi[0]],
        points_to_calc[below_fermi[1]]
    ])
    E_fermi_sea = np.concatenate([
        energies[0][below_fermi[0]],
        energies[1][below_fermi[1]]
    ])
    q_rands = 0.1 * np.random.uniform(-1, 1, size=q_fermi_sea.shape)
    e_rands = 1.0 * np.random.uniform(-1, 1, size=E_fermi_sea.shape)

    ax.axhline(y=fermi_level, color='k', linestyle='--', label='fermi_level')
    ax.scatter(q_fermi_sea + q_rands, E_fermi_sea + e_rands, color='blueviolet', s=4, alpha=0.4)
else:
    rands = np.random.uniform(0, 1, size=fermi_dirac.shape)
    selected = rands <= fermi_dirac
    _qq = np.concatenate([
        points_to_calc[selected[0]],
        points_to_calc[selected[1]]
    ])
    _ee = np.concatenate([
        energies[0][selected[0]],
        energies[1][selected[1]]
    ])
    
    
    ax.axhline(y=mu, color='k', linestyle='--', label='$\\mu$')
    ax.scatter(_qq, _ee, color='blueviolet', s=4, alpha=0.4)
    
ax.legend()

plt.tight_layout()
plt.savefig("band.png", dpi=300, transparent=True)
plt.show()
```

```python
n_lowers = fermi_dirac[0, :]
n_uppers = fermi_dirac[1, :]

alphas_col = alphas[:, np.newaxis]
betas_col = betas[:, np.newaxis]
n_lowers_col = n_lowers[:, np.newaxis]
n_uppers_col = n_uppers[:, np.newaxis]

mat1 = ((alphas_col *  betas) ** 2) * n_uppers_col * n_uppers \
     + (( betas_col *  betas) ** 2) * n_lowers_col * n_uppers \
     + ((alphas_col * alphas) ** 2) * n_uppers_col * n_lowers \
     + (( betas_col * alphas) ** 2) * n_lowers_col * n_lowers
      
mat2 = -( betas_col * alphas * alphas_col *  betas) * n_uppers_col * n_uppers \
     +  (alphas_col * alphas *  betas_col *  betas) * n_lowers_col * n_uppers \
     +  ( betas_col *  betas * alphas_col * alphas) * n_uppers_col * n_lowers \
     -  (alphas_col *  betas *  betas_col * alphas) * n_lowers_col * n_lowers
      
corr_mat = mat1 + mat2
```

```python
def calc_frac(num, den, eps=1e-20):
    if eps is None:
        return num / den
    return np.where(np.abs(den) > eps, num / den, 0.0)
```

```python
initial_up = (alphas ** 2) * n_uppers + (betas ** 2) * n_lowers
initial_down = (betas ** 2) * n_uppers + (alphas ** 2) * n_lowers
initial_total = initial_up + initial_down

# -dn/dt
loss_rate_up = (gamma / L) * np.sum(corr_mat, axis=1) # sum over p
loss_rate_down = (gamma / L) * np.sum(corr_mat.T, axis=1) # sum over p
loss_rate_total = loss_rate_up + loss_rate_down

# -
loss_fraction_up = calc_frac(loss_rate_up, initial_up)
loss_fraction_down = calc_frac(loss_rate_down, initial_down)
loss_fraction_total = calc_frac(loss_rate_total, initial_total)
```

```python
size = initial_up.shape[0]
size_canon = size + 2 * m

points_to_calc_canon = np.linspace(q_left - 1, q_right + 1, (q_right - q_left + 2) * m + 1)

initial_up_canon = np.zeros((size_canon))
initial_up_canon[:-2 * m] = initial_up
initial_down_canon = np.zeros((size_canon))
initial_down_canon[2 * m:] = initial_down
initial_total_canon = initial_up_canon + initial_down_canon

# -dn/dt
loss_rate_up_canon = np.zeros((size_canon))
loss_rate_up_canon[:-2 * m] = loss_rate_up
loss_rate_down_canon = np.zeros((size_canon))
loss_rate_down_canon[2 * m:] = loss_rate_down
loss_rate_total_canon = loss_rate_up_canon + loss_rate_down_canon

loss_fraction_up_canon = calc_frac(loss_rate_up_canon, initial_up_canon)
loss_fraction_down_canon = calc_frac(loss_rate_down_canon, initial_down_canon)
loss_fraction_total_canon = calc_frac(loss_rate_total_canon, initial_total_canon)
```

```python
plt.plot(points_to_calc, loss_fraction_up, color='tomato', label="$\\uparrow$")
plt.plot(points_to_calc, loss_fraction_down, color='cornflowerblue', label="$\\downarrow$")
plt.plot(points_to_calc, loss_fraction_total, color='purple', label="$\\uparrow+\\downarrow$")
plt.legend()
plt.ylim(-0.05, 0.5)
plt.xlim(-5, 5)
plt.xlabel("$q_x/k_r$")
plt.ylabel("Loss Fraction $-(1/n_0(q_x))(d\\langle n(q_x)\\rangle/dt)$")
plt.axhline(y=0, color='k', linestyle='-', linewidth=1)
plt.grid(True, alpha=0.5)
plt.show()
```
