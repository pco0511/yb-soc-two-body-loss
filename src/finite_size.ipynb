```python
import numpy as np
from scipy.stats import binom
import matplotlib.pyplot as plt

def max_loss(n):
  """제공된 max_loss 계산 함수"""
  k = np.arange(n + 1)
  pmf = binom.pmf(k, n, 1/3)
  # 손실을 나타내는 부분: min(k, n-k) * 2
  loss = np.minimum(k, n - k) * 2
  # 기댓값 계산 후 n으로 정규화
  return np.sum(loss * pmf) / n

# n 값을 1부터 30까지 생성
n_values = np.arange(1, 31)

# 각 n 값에 대한 max_loss 계산
loss_values = [max_loss(n) for n in n_values]

# 스캐터 플롯 생성
plt.figure(figsize=(6, 4)) # 플롯 크기 조절

plt.scatter(n_values, loss_values, s=40)

# 플롯 제목 및 라벨 추가
plt.title('Finite Size Effect', fontsize=16)
plt.xlabel('# of particles', fontsize=12)
plt.ylabel('mean maximum loss', fontsize=12)
plt.xticks(np.arange(0, 31, 2)) # x축 눈금 설정 (0부터 30까지 2 간격)

# 플롯 보여주기
plt.show()
```
