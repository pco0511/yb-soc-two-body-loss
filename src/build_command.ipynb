```python
def make_command(
    *,
    python_file_name: str="src/qjm_lindblad_base.py",
    group_name: str="temp",
    run_name_prefix: str="fig3mixture",
    initial_state: str="mixture",
    n_momentum_points: int=13,
    n_particles: int=6,
    n_savesteps: int=51,
    gamma: float=0.4,
    T2: float=0.8,
    sim_time: float=2.0,
    nh: bool=False,
    seed: int=0
):
    commands = [
        "uv", "run", python_file_name, 
        "--group_name", group_name,
        "--run_name", f"{run_name_prefix}_{{default}}", 
        "--initial_state", initial_state,
        "--n_momentum_points", f"{n_momentum_points}", 
        "--n_particles", f"{n_particles}", 
        "--n_savesteps", f"{n_savesteps}", 
        # "--jumps_per_step", "4",
        "--gamma", f"{gamma:.4f}",
        "--T2", f"{T2:.4f}",
        "--sim_time", f"{sim_time:.4f}",
        "--seed", f"{seed}"
    ]
    if nh:
        commands.append("--nh")
    return commands
```

```python
print(" ".join(make_command(group_name="temp", run_name_prefix="nh_test", initial_state="mixture", n_savesteps=51, sim_time=2.0, gamma=0.43, T2=1.5, nh=True)))
```
