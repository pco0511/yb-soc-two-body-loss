```python
import ybsoc
import ybsoc_new
import numpy as np
```

```python
k0 = 0.5
L = 2 * np.pi / k0
n_momentum_points = 9
N = 2

hbar = 1.

m_Yb = 1.

# q = 1.
# delta = 1.5486
# omega_R = 1.0585

k_r = 1.
delta = 4.
omega_R = 0 # 3.5

gamma = 0.32
d = 2 * (2 * np.pi) / n_momentum_points

system = ybsoc.YbSOC2bodyLoss(
    d=d,
    lengths=[n_momentum_points,],
    n_particle=N,
    hbar=hbar,
    q=k_r,
    m_Yb=m_Yb,
    delta=delta,
    omega_R=omega_R,
    gamma=gamma,
    array_type='numpy'
)

new_system = ybsoc_new.YbSOC2bodyLoss(
    L=L,
    n_mometum_points=n_momentum_points,
    n_particle=N,
    hbar=hbar,
    k_r=k_r,
    m_Yb=m_Yb,
    delta=delta,
    omega_R=omega_R,
    gamma=gamma,
)

system.print_info()
```

```python
dense = system.dense_hamiltonian(kinetic=True, interaction=False)
```

```python
dense_new = new_system.dense_hamiltonian(kinetic=True, interaction=False)
```
