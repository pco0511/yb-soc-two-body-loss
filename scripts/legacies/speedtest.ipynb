```python
import os


import jax

jax.config.update("jax_enable_x64", True)
jax.config.update('jax_use_magma', 'on')

import jax.numpy as jnp
from jax.extend.backend import get_backend
import numpy as np
import tqdm


import matplotlib.pyplot as plt


from ybsoc import *
```

```python
system = YbSOCSystem(
    d = 1,
    lengths = [8, ],
    n_particle = 6,

    hbar = 1,
    q = 0.03,
    m_Yb = 1,
    delta = 0.08,
    omega_R = 0.2,
    gamma = 0.1
)
hamiltonian = system.dense_hamiltonian()
```

```python
hamiltonian_jaxarray = jnp.array(hamiltonian)
hamiltonian_numpyarray = np.array(hamiltonian)
hamiltonian_torchtensor = torch.tensor(hamiltonian_numpyarray).cuda()
```
