```python
import itertools
import collections
import functools

import tqdm

from opalg import lattice
```

```python
L = [15]
s = [1, -1]
n = 5
```

```python
single_states = list(itertools.product(lattice.multi_indices(L), s))
```

```python
counter = collections.defaultdict(list)
```

```python
for multi_state in tqdm.tqdm(list(itertools.combinations(single_states, n))):
    k_total = functools.reduce(functools.partial(lattice.add_multi_indices, lengths=L), (momentum for momentum, _ in multi_state))
    counter[k_total].append(multi_state)
```

```python
subdims = {key:len(states) for key, states in counter.items()}
```
