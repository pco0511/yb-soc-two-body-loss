```python
from diffrax import diffeqsolve, ODETerm, Dopri5, SaveAt
import jax.numpy as jnp

def f(t, y, args):
    return -y

term = ODETerm(f)
solver = Dopri5()
saveat = SaveAt(ts=jnp.linspace(0, 1, 21))
y0 = jnp.array([2., 3.])
solution = diffeqsolve(term, solver, t0=0, t1=1, dt0=0.1, y0=y0, saveat=saveat)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(solution.ts, solution.ys)
plt.show()
```
